# Project Bootstrap Notebook
Initialise the full MSc thesis project directory structure in a reproducible way.
#### Rationale:
1. Ensure datasets, code, and outputs are placed in a consistent, modular structure.
2. Keep raw data immutable (`data/`) and separate from analysis and outputs.
3. Facilitate clean imports by treating `src/` as Python package.
4. Provide examiners and collaborators with a one-step reproducible setup, avoiding manual folder creation.

> This notebook is intentionally kept in the project **root**, since it creates the `notebooks/` folder itself.
> It can be re-run at any time without harm (idempotent).

```python
# Create the base project structure safely and idempotently.
# This ensures reproducibility: examiners or colaborators can re-run this to set up the same workspace.

from pathlib import Path

# Define the project root (this notebook is already inside it).
PROJECT_ROOT = Path(".").resolve()

# Key subdirectories to initialise.
dirs = [
    PROJECT_ROOT / "data" / "chexpert",
    PROJECT_ROOT / "data" / "tbx11k",
    PROJECT_ROOT / "notebook",
    PROJECT_ROOT / "src",
    PROJECT_ROOT / "experiments",
    PROJECT_ROOT / "reports" / "figures",
    PROJECT_ROOT / "reports" / "tables",
    PROJECT_ROOT / "reports" / "thesis",
    PROJECT_ROOT / "reports" / "Documentation",
]

# Create al directories.

for d in dirs:
    d.mkdir(parents=True, exist_ok=True)

print(f'Project structure initialised at: {PROJECT_ROOT}')
for d in dirs:
    print(".", d.relative_to(PROJECT_ROOT))
```
